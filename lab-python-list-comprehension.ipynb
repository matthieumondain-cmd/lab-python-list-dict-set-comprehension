## Exercise: Managing Customer Orders Optimized with Comprehension

````python
inventory = {} 
products=["t-shirt", "mug", "hat", "book", "keychain"] #that takes products as a parameter
def initialize_inventory(products): #Define a function named initialize_inventory #Inside the function
    inventory = {product:(int(input(f"How many {product} did you count :"))) for product in products}   
    return inventory
inventory =  initialize_inventory(products)
print(inventory)

def get_customer_orders(): #Define a function named get_customer_orders that takes no parameters.
    num_orders= int(input("how many different articles do you want to order?"))
    customer_orders ={}
    for _ in range(num_orders) :
        product = input("Which articles do you want to order?").lower()
        quantity = int(input("How many of them you want to order ?"))
        customer_orders[product] = customer_orders.get(product, 0) + quantity
    return customer_orders
customer_orders=get_customer_orders()
print(customer_orders)
type(customer_orders)


def calculate_order_statistics(customer_orders , products) : #Define a function named calculate_order_statistics that takes customer_orders and products as parameters.
    total_products_ordered = sum(customer_orders.values())
    unique_products_ordered = len(customer_orders)
    percentage_unique=(unique_products_ordered/len(products))*100      
    return total_products_ordered, percentage_unique


total_ordered, percent_unique = calculate_order_statistics(customer_orders, products)
order_statistics = total_ordered, percent_unique
print (order_statistics)
print(f"total products ordered : {total_ordered}")
print(f"percentage unique : {percent_unique}%")
type(order_statistics)
    # Inside the function, implement the code for calculating the order statistics (total products ordered, and percentage of unique products ordered). The function should return these values.
#Exercise 5
def print_order_statistics(order_statistics): #Define a function named print_order_statistics that takes order_statistics as a parameter.
    print (f"total products ordered, percentage unique: {order_statistics}")
order_statistics = total_ordered, percent_unique
print_order_statistics(order_statistics)


"""
Add a new function to calculate the total price of the customer order.
For each product in customer_orders, prompt the user to enter the price of that product.
Use comprehension to calculate the total price.
Note: assume that the user can only have 1 unit of each product.
"""
def price_order():
    total_price=sum (float(input(f"what's the price for a {product} ?")) for product in customer_orders)
    return total_price
total = price_order()
print(f"Total price: {total}")
"""
Modify the update_inventory function to remove the product from the inventory if its quantity becomes zero after fulfilling the customer orders.
Use comprehension to filter out the products with a quantity of zero from the inventory.
"""



def update_inventory(customer_orders , inventory) :
    for product, quantity in customer_orders.items() :
        if product in inventory and inventory[product]>= quantity :
            inventory[product] -= quantity
            
        else :
            print(f"Oops, there is not {product} available in stock !")
    inventory = {p : q for p, q in inventory.items() if q >0} 
    return inventory

inventory_updated = update_inventory(customer_orders , inventory)
print("Updated inventory",inventory_updated)  






````
